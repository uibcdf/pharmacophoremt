```python
from IPython.display import display
import molsysviewer as mv
from pharmacophoremt import Pharmacophore
from pharmacophoremt.interaction_site import HBAcceptorSphere, HydrophobicSphere
from molsysviewer_pharmacophoremt import ensure_runtime
model = Pharmacophore()
model.interaction_sites.extend([
    HBAcceptorSphere(center="[0, 0, 0] nm", radius="0.4 nm"),
    HydrophobicSphere(center="[1.2, 0, 0] nm", radius="0.35 nm"),
])
model.n_interaction_sites = 2
view = mv.demo["dialanine"]
view.addons.enable("pharmacophoremt")
runtime = ensure_runtime(view)
runtime.pharmacophore = model
view.shapes.add_sphere(center="[0, 0.8, 0] nm", radius="0.15 nm", color="#FFFFFF", tag="unrelated-control")
view.set_workspace_panel(workspace="pharmacophoremt", panel="pharmacophore")
display(view)
```
